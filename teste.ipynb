```python
import pandas as pd
import datetime as dt
import numpy as np
```

```python
df = pd.read_excel("CustomerLineItems (25).xlsx")
```

```python
df['data_hoje'] = dt.datetime.now().date()
```

```python
df['data_hoje'] = pd.to_datetime(df['data_hoje'])
df['Data vencimento líq.'] = pd.to_datetime(df['Data vencimento líq.'])

df['Contagem_dias'] = (
    df['data_hoje'] - df['Data vencimento líq.']
).dt.days
```

```python
condicoes = [
    (df['Contagem_dias'] <= 0),
    (df['Contagem_dias'] > 0) & (df['Contagem_dias'] <= 30),
    (df['Contagem_dias'] > 30) & (df['Contagem_dias'] <= 60),
    (df['Contagem_dias'] > 60) & (df['Contagem_dias'] <= 90),
    (df['Contagem_dias'] > 90) & (df['Contagem_dias'] <= 10000)
]

resultado = ["Não considerado", "A Vencer", "Vencido 30 dias", "Vencido 60 dias", "Vencido 90 dias"]

df['Contagem_Prazo']= np.select(condicoes,resultado, default="Descondiderado")
```
